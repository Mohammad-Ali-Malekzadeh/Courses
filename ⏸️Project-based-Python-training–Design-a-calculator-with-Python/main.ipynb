# Settings

## root = Tk()
<p style="direction: rtl">
این کد، پنجرهٔ اصلی (Main Window) برنامه‌های گرافیکی tkinter را می‌سازد.<br>
برای درک بهتر، بیایید خط‌به‌خط بررسی کنیم:<br>
۱. Tk() چیست؟<br>
Tk یک کلاس در ماژول tkinter است. وقتی آن را صدا می‌زنید (Tk())، یک شیء ساخته می‌شود که نمایانگر خودِ برنامه (application) و پنجرهٔ اصلی آن است.<br>
این کار چند کار اساسی انجام می‌دهد:<br>
یک پنجرهٔ خالی با عنوان پیش‌فرض ("tk") و اندازهٔ کوچک باز می‌کند.<br>
یک حلقه‌ی رویداد (Event Loop) راه‌اندازی می‌کند (اگر بعداً mainloop() صدا زده شود) تا منتظر کلیک‌ها و ورودی‌های کاربر بماند.<br>
به عنوان ظرف (Container) اصلی عمل می‌کند؛ یعنی تمام دکمه‌ها، برچسب‌ها و جعبه‌های متنی دیگر، داخل همین پنجره قرار می‌گیرند.<br>
۲. root = Tk() یعنی چه؟<br>
با این کار، شیء ساخته‌شده را در متغیری به نام root ذخیره می‌کنید. این متغیر حالا به همه‌ی برنامه دسترسی دارد و شما می‌توانید با آن کار کنید
</p>

```python
root = Tk()
root.title("Calculator")  # تایتل برنامه مون
root.geometry("400x200")  # طول و عرض برنامه رو تنظیم میکنه
root.resizable(width=False, height=False)  # جلوی ریسایز کردن رو میگیره
```

<p style="direction: rtl">
متد configure() چیست؟<br>
configure یک متد قدرتمند در tkinter است که به شما اجازه می‌دهد ویژگی‌های ظاهری یک ویجت را بعد از ساخته شدن، تغییر دهید. معادل مستقیم آن، تنظیم ویژگی‌ها در زمان ساخت است:
در اکثر ویجت‌های tkinter، هر دو کلمه bg و background معادل هم هستند و یکی را به جای دیگری هم می‌توانید استفاده کنید.
</p>
<div>
<pre>
<code class="language-python">
# این دو خط کاملاً معادل هستند:
root = tk.Tk(bg="gray")        # تنظیم در زمان ساخت
root.configure(bg="gray")      # تنظیم بعد از ساخت (تغییر ویژگی)
</code>
</pre>
</div>


    

```python
color = "gray"
root.configure(bg=color)  
```
